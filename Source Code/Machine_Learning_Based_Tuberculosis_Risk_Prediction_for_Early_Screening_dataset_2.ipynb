```python
# ============================================================
# DATASET 2 — LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np

# Load Dataset 2
df2 = pd.read_csv("tb_diagnosis.csv")

print("=" * 60)
print("DATASET 2 — TB DIAGNOSIS DATASET")
print("=" * 60)

print("\nDataset Shape:")
print(df2.shape)

print("\nDataset Columns:")
print(df2.columns.tolist())

print("\nFirst 5 Rows:")
display(df2.head())
```

```python
# ============================================================
# DATASET 2 — DATA UNDERSTANDING
# ============================================================

print("=" * 60)
print("DATASET 2 — DATA UNDERSTANDING")
print("=" * 60)

# Dataset information
print("\n1. Dataset Information:")
print("-" * 40)
df2.info()

# Statistical summary
print("\n2. Statistical Summary:")
print("-" * 40)
display(df2.describe(include="all").T)

# Missing values
print("\n3. Missing Values:")
print("-" * 40)
missing_values_2 = df2.isnull().sum()

display(
    missing_values_2.to_frame(
        name="Missing Values"
    )
)

# Duplicate records
print("\n4. Duplicate Records:")
print("-" * 40)
print("Number of duplicate rows:", df2.duplicated().sum())

# Target distribution
print("\n5. Target Variable Distribution:")
print("-" * 40)
print(df2["tb"].value_counts())

print("\nTarget Variable Percentage:")
display(
    (df2["tb"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# ============================================================
# DATASET 2 — FEATURE AND TARGET SEPARATION
# ============================================================

print("=" * 60)
print("DATASET 2 — FEATURE AND TARGET SEPARATION")
print("=" * 60)

# Target variable
y2 = df2["tb"]

# Input features
X2 = df2.drop("tb", axis=1)

print("\nTarget variable:")
print("tb")

print("\nTarget shape:", y2.shape)
print("Feature shape:", X2.shape)

print("\nFeature columns:")
print(X2.columns.tolist())

print("\nTarget distribution:")
print(y2.value_counts())

print("\nTarget distribution (%):")
display(
    (y2.value_counts(normalize=True) * 100).round(2)
)
```

```python
# ============================================================
# DATASET 2 — CATEGORICAL ENCODING
# ============================================================

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

print("=" * 60)
print("DATASET 2 — CATEGORICAL FEATURE ENCODING")
print("=" * 60)

# Identify categorical and numerical columns
categorical_features_2 = ["age_group", "country"]

numerical_features_2 = [
    "hiv_pos",
    "diabetes",
    "ever_smoke",
    "past_tb",
    "male",
    "hs_less",
    "two_weeks_symp",
    "num_symptoms"
]

print("\nCategorical features:")
print(categorical_features_2)

print("\nNumerical features:")
print(numerical_features_2)

# Preprocessing
preprocessor_2 = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features_2
        ),
        (
            "numerical",
            "passthrough",
            numerical_features_2
        )
    ]
)

# Apply preprocessing
X2_processed = preprocessor_2.fit_transform(X2)

print("\nOriginal feature shape:", X2.shape)
print("Processed feature shape:", X2_processed.shape)

print("\nCategorical encoding completed successfully.")
```

```python
# ============================================================
# DATASET 2 — TRAIN TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

print("=" * 60)
print("DATASET 2 — TRAIN TEST SPLIT")
print("=" * 60)

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2_processed,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

print("\nTraining data:")
print("X2_train:", X2_train.shape)
print("y2_train:", y2_train.shape)

print("\nTesting data:")
print("X2_test:", X2_test.shape)
print("y2_test:", y2_test.shape)

print("\nTrain/Test split completed successfully.")
```

```python
# ============================================================
# DATASET 2 — TRAIN ALL MACHINE LEARNING MODELS
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ------------------------------------------------------------
# 1. Define all models
# ------------------------------------------------------------

models_2 = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=100,
        max_depth=4,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="logloss"
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=100,
        random_state=42
    ),

    "Gaussian Naive Bayes": GaussianNB(),

    "Neural Network": MLPClassifier(
        hidden_layer_sizes=(100,),
        max_iter=500,
        random_state=42
    )
}

print("=" * 70)
print("DATASET 2 — TRAINING ALL MODELS")
print("=" * 70)

# ------------------------------------------------------------
# 2. Train and evaluate each model
# ------------------------------------------------------------

results_2 = []
trained_models_2 = {}

for name, model in models_2.items():

    print(f"\nTraining {name}...")

    # Train model
    model.fit(X2_train, y2_train)

    # Store trained model
    trained_models_2[name] = model

    # Predictions
    y_pred_temp = model.predict(X2_test)

    # Probability / decision scores
    if hasattr(model, "predict_proba"):
        y_prob_temp = model.predict_proba(X2_test)[:, 1]
    elif hasattr(model, "decision_function"):
        y_prob_temp = model.decision_function(X2_test)
    else:
        y_prob_temp = y_pred_temp

    # Metrics
    accuracy_temp = accuracy_score(y2_test, y_pred_temp)
    precision_temp = precision_score(
        y2_test, y_pred_temp, zero_division=0
    )
    recall_temp = recall_score(
        y2_test, y_pred_temp, zero_division=0
    )
    f1_temp = f1_score(
        y2_test, y_pred_temp, zero_division=0
    )
    roc_auc_temp = roc_auc_score(
        y2_test, y_prob_temp
    )

    results_2.append({
        "Model": name,
        "Accuracy": accuracy_temp,
        "Precision": precision_temp,
        "Recall": recall_temp,
        "F1-Score": f1_temp,
        "ROC-AUC": roc_auc_temp
    })

    print(f"✓ {name} completed")

# ------------------------------------------------------------
# 3. Create comparison table
# ------------------------------------------------------------

results_df_2 = pd.DataFrame(results_2)

# Sort by Recall because TB screening prioritizes
# identifying suspected cases
results_df_2 = results_df_2.sort_values(
    by="Recall",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------
# 4. Display results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET 2 — MODEL PERFORMANCE COMPARISON")
print("=" * 70)

display(results_df_2.round(4))
```

```python
# ============================================================
# DATASET 2 — SELECT BEST MODEL
# ============================================================

# Select model with highest Recall
best_model_name_2 = results_df_2.loc[0, "Model"]

best_model_2 = trained_models_2[best_model_name_2]

print("=" * 70)
print("DATASET 2 — BEST MODEL SELECTION")
print("=" * 70)

print(f"\nBest Model: {best_model_name_2}")

print("\nReason:")
print("The model with the highest Recall is selected because")
print("TB screening prioritizes identifying suspected TB cases.")

print("\nBest Model Performance:")
display(
    results_df_2[
        results_df_2["Model"] == best_model_name_2
    ].round(4)
)
```

```python
# ============================================================
# DATASET 2 — FINAL XGBOOST MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve
)

import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Predictions
# ------------------------------------------------------------

y_pred_2 = best_model_2.predict(X2_test)

# Probability of TB Suspected class
y_prob_2 = best_model_2.predict_proba(X2_test)[:, 1]

# ------------------------------------------------------------
# 2. Calculate metrics
# ------------------------------------------------------------

accuracy_2 = accuracy_score(y2_test, y_pred_2)

precision_2 = precision_score(
    y2_test,
    y_pred_2,
    zero_division=0
)

recall_2 = recall_score(
    y2_test,
    y_pred_2,
    zero_division=0
)

f1_2 = f1_score(
    y2_test,
    y_pred_2,
    zero_division=0
)

roc_auc_2 = roc_auc_score(
    y2_test,
    y_prob_2
)

# ------------------------------------------------------------
# 3. Display final performance
# ------------------------------------------------------------

print("=" * 70)
print("DATASET 2 — FINAL XGBOOST MODEL PERFORMANCE")
print("=" * 70)

print(f"Accuracy  : {accuracy_2:.4f} ({accuracy_2*100:.2f}%)")
print(f"Precision : {precision_2:.4f} ({precision_2*100:.2f}%)")
print(f"Recall    : {recall_2:.4f} ({recall_2*100:.2f}%)")
print(f"F1-Score  : {f1_2:.4f} ({f1_2*100:.2f}%)")
print(f"ROC-AUC   : {roc_auc_2:.4f} ({roc_auc_2*100:.2f}%)")


# ------------------------------------------------------------
# 4. Classification Report
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y2_test,
        y_pred_2,
        target_names=[
            "Not Suspected",
            "TB Suspected"
        ],
        zero_division=0
    )
)


# ------------------------------------------------------------
# 5. Confusion Matrix
# ------------------------------------------------------------

cm_2 = confusion_matrix(
    y2_test,
    y_pred_2
)

print("\n" + "=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

print(cm_2)

plt.figure(figsize=(6, 5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_2,
    display_labels=[
        "Not Suspected",
        "TB Suspected"
    ]
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Dataset 2 — XGBoost Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. ROC Curve
# ------------------------------------------------------------

fpr_2, tpr_2, thresholds_2 = roc_curve(
    y2_test,
    y_prob_2
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr_2,
    tpr_2,
    linewidth=2,
    label=f"XGBoost (AUC = {roc_auc_2:.2f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Dataset 2 — XGBoost ROC Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nDataset 2 final XGBoost evaluation completed successfully.")
```

```python
# ============================================================
# DATASET 2 — XGBOOST HYPERPARAMETER TUNING
# ============================================================

from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ------------------------------------------------------------
# 1. Define XGBoost model
# ------------------------------------------------------------

xgb_base = XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

# ------------------------------------------------------------
# 2. Define parameters to test
# ------------------------------------------------------------

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [3, 4, 5],
    "learning_rate": [0.05, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}

# ------------------------------------------------------------
# 3. Grid Search
# ------------------------------------------------------------

print("Starting XGBoost hyperparameter tuning...")
print("Please wait...")

grid_search_2 = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_grid,
    scoring="recall",
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_search_2.fit(X2_train, y2_train)

# ------------------------------------------------------------
# 4. Best parameters
# ------------------------------------------------------------

best_xgb_2 = grid_search_2.best_estimator_

print("\n" + "=" * 70)
print("XGBOOST TUNING COMPLETED")
print("=" * 70)

print("\nBest Parameters:")
print(grid_search_2.best_params_)

print(f"\nBest Cross-Validation Recall: "
      f"{grid_search_2.best_score_:.4f}")
```

```python
# ============================================================
# DATASET 2 — EVALUATE TUNED XGBOOST
# ============================================================

# Predictions
y_pred_tuned_2 = best_xgb_2.predict(X2_test)

# Probabilities
y_prob_tuned_2 = best_xgb_2.predict_proba(X2_test)[:, 1]

# Metrics
accuracy_tuned_2 = accuracy_score(
    y2_test,
    y_pred_tuned_2
)

precision_tuned_2 = precision_score(
    y2_test,
    y_pred_tuned_2,
    zero_division=0
)

recall_tuned_2 = recall_score(
    y2_test,
    y_pred_tuned_2,
    zero_division=0
)

f1_tuned_2 = f1_score(
    y2_test,
    y_pred_tuned_2,
    zero_division=0
)

roc_auc_tuned_2 = roc_auc_score(
    y2_test,
    y_prob_tuned_2
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("DATASET 2 — TUNED XGBOOST PERFORMANCE")
print("=" * 70)

print(f"Accuracy  : {accuracy_tuned_2:.4f} "
      f"({accuracy_tuned_2*100:.2f}%)")

print(f"Precision : {precision_tuned_2:.4f} "
      f"({precision_tuned_2*100:.2f}%)")

print(f"Recall    : {recall_tuned_2:.4f} "
      f"({recall_tuned_2*100:.2f}%)")

print(f"F1-Score  : {f1_tuned_2:.4f} "
      f"({f1_tuned_2*100:.2f}%)")

print(f"ROC-AUC   : {roc_auc_tuned_2:.4f} "
      f"({roc_auc_tuned_2*100:.2f}%)")
```

```python
# ============================================================
# DATASET 2 — BASELINE VS TUNED XGBOOST
# ============================================================

comparison_xgb_2 = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ],

    "Baseline XGBoost": [
        accuracy_2,
        precision_2,
        recall_2,
        f1_2,
        roc_auc_2
    ],

    "Tuned XGBoost": [
        accuracy_tuned_2,
        precision_tuned_2,
        recall_tuned_2,
        f1_tuned_2,
        roc_auc_tuned_2
    ]
})

display(
    comparison_xgb_2.round(4)
)
```

```python
# ============================================================
# DATASET 2 — XGBOOST THRESHOLD ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Use probabilities from the tuned XGBoost model
threshold_probabilities_2 = best_xgb_2.predict_proba(X2_test)[:, 1]

threshold_results_2 = []

# Thresholds to test
thresholds = [0.50, 0.45, 0.40, 0.35, 0.30]

for threshold in thresholds:

    # Convert probability into class prediction
    y_pred_threshold = (
        threshold_probabilities_2 >= threshold
    ).astype(int)

    # Confusion matrix
    tn, fp, fn, tp = confusion_matrix(
        y2_test,
        y_pred_threshold
    ).ravel()

    # Metrics
    accuracy = accuracy_score(
        y2_test,
        y_pred_threshold
    )

    precision = precision_score(
        y2_test,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y2_test,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y2_test,
        y_pred_threshold,
        zero_division=0
    )

    threshold_results_2.append({
        "Threshold": threshold,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

# Create table
threshold_df_2 = pd.DataFrame(
    threshold_results_2
)

print("=" * 80)
print("DATASET 2 — XGBOOST THRESHOLD ANALYSIS")
print("=" * 80)

display(
    threshold_df_2.round(4)
)
```

```python
# ============================================================
# DATASET 2 — FINAL THRESHOLD SELECTION
# ============================================================

FINAL_THRESHOLD_2 = 0.40

# Generate final predictions using selected threshold
y_pred_final_2 = (
    threshold_probabilities_2 >= FINAL_THRESHOLD_2
).astype(int)

# Calculate confusion matrix
tn_final_2, fp_final_2, fn_final_2, tp_final_2 = confusion_matrix(
    y2_test,
    y_pred_final_2
).ravel()

# Calculate final metrics
accuracy_final_2 = accuracy_score(
    y2_test,
    y_pred_final_2
)

precision_final_2 = precision_score(
    y2_test,
    y_pred_final_2,
    zero_division=0
)

recall_final_2 = recall_score(
    y2_test,
    y_pred_final_2,
    zero_division=0
)

f1_final_2 = f1_score(
    y2_test,
    y_pred_final_2,
    zero_division=0
)

print("=" * 70)
print("DATASET 2 — FINAL XGBOOST MODEL")
print("=" * 70)

print(f"\nSelected Threshold : {FINAL_THRESHOLD_2}")

print("\nConfusion Matrix Values:")
print(f"True Negative  (TN): {tn_final_2}")
print(f"False Positive (FP): {fp_final_2}")
print(f"False Negative (FN): {fn_final_2}")
print(f"True Positive  (TP): {tp_final_2}")

print("\nFinal Performance:")
print(f"Accuracy  : {accuracy_final_2:.4f} ({accuracy_final_2*100:.2f}%)")
print(f"Precision : {precision_final_2:.4f} ({precision_final_2*100:.2f}%)")
print(f"Recall    : {recall_final_2:.4f} ({recall_final_2*100:.2f}%)")
print(f"F1-Score  : {f1_final_2:.4f} ({f1_final_2*100:.2f}%)")

print("\nReason for selecting threshold 0.40:")
print("It provides a better balance between TB case detection")
print("and false-positive predictions.")
```

```python
# ============================================================
# DATASET 2 — FINAL CONFUSION MATRIX
# ============================================================

cm_final_2 = confusion_matrix(
    y2_test,
    y_pred_final_2
)

plt.figure(figsize=(6, 5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_final_2,
    display_labels=[
        "Not Suspected",
        "TB Suspected"
    ]
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title(
    "Dataset 2 — Final XGBoost Confusion Matrix\n"
    "Threshold = 0.40"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# DATASET 2 — FINAL ROC CURVE
# ============================================================

from sklearn.metrics import roc_curve, roc_auc_score

fpr_final_2, tpr_final_2, _ = roc_curve(
    y2_test,
    threshold_probabilities_2
)

roc_auc_final_2 = roc_auc_score(
    y2_test,
    threshold_probabilities_2
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr_final_2,
    tpr_final_2,
    linewidth=2,
    label=f"XGBoost (AUC = {roc_auc_final_2:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "Dataset 2 — Final XGBoost ROC Curve"
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Final ROC-AUC: {roc_auc_final_2:.4f}")
```
